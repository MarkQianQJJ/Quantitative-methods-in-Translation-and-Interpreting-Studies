```python
# Install first:
# pip install sacrebleu

import sacrebleu
```

```python
# -----------------------------
# 1. Example translations
# -----------------------------

# System outputs / machine translations
hypotheses = [
    "The cat is on the mat.",
    "There is a book on the table.",
    "Machine translation is useful."
]

# Reference translations
# SacreBLEU expects references as a list of reference sets.
# For one reference per sentence, use: [references]
references = [
    "The cat sits on the mat.",
    "There is a book on the desk.",
    "Machine translation is helpful."
]
```

```python
# -----------------------------
# 2. Corpus-level SacreBLEU
# -----------------------------

bleu = sacrebleu.corpus_bleu(
    hypotheses,
    [references]
)

print("BLEU score:", bleu.score)
print("Full BLEU output:")
print(bleu)
```

```python
# -----------------------------
# 3. Sentence-level SacreBLEU
# -----------------------------

sentence_hyp = "The cat is on the mat."
sentence_ref = "The cat sits on the mat."

sent_bleu = sacrebleu.sentence_bleu(
    sentence_hyp,
    [sentence_ref]
)

print("\nSentence BLEU score:", sent_bleu.score)
print("Full sentence BLEU output:")
print(sent_bleu)
```
