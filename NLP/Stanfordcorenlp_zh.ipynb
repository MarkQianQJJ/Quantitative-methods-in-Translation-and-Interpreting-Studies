```python
from stanfordcorenlp import StanfordCoreNLP
```

```python
zh_model = StanfordCoreNLP(r'D:/vscode/stanford-corenlp-full-2018-10-05',lang='zh')
```

```python
zh_sentence='相对地, 中国的佛教, 也不全同于印度或其他国家的佛教'
```

```python
print('Tokenize:',zh_model.word_tokenize(zh_sentence))
```

```python
print('POS tag:',zh_model.pos_tag(zh_sentence))
```

```python
print('Named entity:',zh_model.ner(zh_sentence))
```

```python
print('Consituency parsing:',zh_model.parse(zh_sentence))
```

```python
print('Dependency parsing:',zh_model.dependency_parse(zh_sentence))
```
