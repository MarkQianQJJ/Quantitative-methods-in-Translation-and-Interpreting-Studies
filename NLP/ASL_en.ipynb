```python
import nltk
import re
```

```python
string = "My father's family name being Pirrip, and my Christian name Philip, my infant tongue could make of both names nothing longer or more explicit than Pip. So, I called myself Pip, and came to be called Pip."
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
sent_tokenizer = nltk.data.load('tokenizers/punkt/english.pickle')
```

```python
sents = sent_tokenizer.tokenize(string)

print("句子列表：", sents)
```

```python
sentence_lengths = []
for sent in sents:
    words = re.findall(r"\b\w+(?:'\w+)?\b", sent)
    sentence_lengths.append(len(words))
```

```python
print("每句长度：", sentence_lengths)
```

```python
avg_len = sum(sentence_lengths) / len(sentence_lengths)
```

```python
print("平均句长：", avg_len)
```
