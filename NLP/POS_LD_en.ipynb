```python
import nltk
import string
#https://www.nltk.org/data.html
```

```python
text = "My father's family name being Pirrip, and my Christian name Philip, my infant tongue could make of both names nothing longer or more explicit than Pip. So, I called myself Pip, and came to be called Pip."
```

```python
string_tokenized = nltk.word_tokenize(text)
```

```python
string_postagged = nltk.pos_tag(string_tokenized)
print(string_postagged)
```

```python
for i in string_postagged:
    print(i[0] + '_' + i[1])
```

```python
content_tags = ('NN', 'VB', 'JJ', 'RB')
```

```python
content_words = [word for word, tag in string_postagged if tag.startswith(content_tags)]
print(content_words)
```

The following code creates a new list called words_only that contains all the words from string_postagged except punctuation marks. The expression loops through each (word, tag) pair in string_postagged, checks whether the word is not in string.punctuation, and keeps it only if it is not a punctuation symbol such as a comma or period. As a result, words_only stores just the actual word tokens from the text, while print(words_only) displays that filtered list.

```python
words_only = [word for word, tag in string_postagged
              if word not in string.punctuation]

print(words_only)
```

```python
lexical_density = len(content_words) / len(words_only)

print("总词数：", len(words_only))
print("实词数：", len(content_words))
print("实词：", content_words)
print("实词密度：", lexical_density)
print("实词密度（百分比）：{:.2%}".format(lexical_density))
```
