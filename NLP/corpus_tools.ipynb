Keyness is calculated using two frequency dictionaries (consisting of raw frequency values). Only effect sizes are reported (p values are arguably not particularly useful for keyness analyses). Keyness calculation options include "log-ratio", "%diff", and "odds-ratio".
https://pypi.org/project/corpus-toolkit/

```python
from corpus_toolkit import corpus_tools as ct

brown_corp = ct.ldcorpus(
    r"D:\vscode\corpus data\corpus_toolkit\brown_single\brown_single"
)
tok_corp = ct.tokenize(brown_corp)
brown_freq = ct.frequency(tok_corp)
ct.head(brown_freq, hits=10)
```

```python
conc_results1 = ct.concord(ct.tokenize(ct.ldcorpus(r"D:\vscode\corpus data\corpus_toolkit\brown_single\brown_single"),lemma = False),["run","ran","running","runs"],nhits = 10)
for x in conc_results1:
	print(x)
```
